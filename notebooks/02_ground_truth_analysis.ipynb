# Amazon ML Challenge 2026: Ground Truth Analysis
Analysis of match cardinality, singleton distribution, target source split (S2 vs S3), and cross-country evaluation.

```python
import json
import pandas as pd

with open('../reports/dataset_profile.json', 'r') as f:
    profile = json.load(f)

gt = profile['ground_truth']
summary = {
    'Total S1 Entities': gt['total_s1'],
    'Singletons (0 matches)': f"{gt['zero_matches_count']} ({gt['zero_matches_pct']}%)",
    'Single Match (1 match)': f"{gt['one_match_count']} ({gt['one_match_pct']}%)",
    'Multi Matches (>1)': f"{gt['multi_matches_count']} ({gt['multi_matches_pct']}%)",
    'Total Matches': gt['total_matches_across_all'],
    'Avg Matches per S1': gt['average_matches_per_s1'],
    'Max Matches for an S1': gt['max_matches_per_s1'],
    'Cross-Country Matches': gt['cross_country_matches']
}
pd.DataFrame(list(summary.items()), columns=['Metric', 'Value'])
```
