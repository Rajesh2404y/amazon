# Amazon ML Challenge 2026: Dataset Analysis & Profiling
This notebook inspects the challenge dataset across Source 1, Source 2, and Source 3 for both Training and Testing splits.

```python
import os
import json
import pandas as pd

# Load precomputed profile
with open('../reports/dataset_profile.json', 'r') as f:
    profile = json.load(f)

print("Ground Truth Summary:")
for k, v in profile['ground_truth'].items():
    print(f"  {k}: {v}")
```

```python
# Display training source statistics
train_df = pd.DataFrame(profile['train_sources']).T[['total_rows', 'unique_entity_ids', 'missing_counts', 'country_distribution']]
train_df
```

```python
# Display test source statistics
test_df = pd.DataFrame(profile['test_sources']).T[['total_rows', 'unique_entity_ids', 'missing_counts', 'country_distribution']]
test_df
```
